```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5,5,100)
y1 = (3-2*x)/3
y2 = (6-4*x)/6
y3 = (3*x-1)/2

fig, ax = plt.subplots()

ax.plot(x,y1, color='r')
ax.plot(x,y2, color='g')
ax.plot(x,y3, color='b')

ax.text(0.5,0.6, '$2x+3y = 3$', color='r')
ax.text(-3,0.5,'$4x+6y = 6$', color='g')
ax.text(1.2,2, '$3x-2y = 1$', color='b')

ax.set_xlim(-4,4)
ax.set_ylim(-2,6)

ax.axvline(color='k',linewidth = 1)
ax.axhline(color='k',linewidth = 1)

ax.set_xticks(list(range(-4,5)))
ax.set_aspect('equal')
ax.grid(True,ls=':')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
matrix = np.random.randint(1, 21, (4, 4))
inp = input("\nEnter a number to search: ")
count = 0;
if inp.isdigit():
  num = int(inp)
  for i in range(0,4):
    for j in range(0,4):
      if(num == matrix[i][j]):
        count += 1;
  if count > 0:
    print(f"Number {num} found {count} times in the matrix.")
  else:
    print(f"Number {num} is not in the matrix.")
  print(matrix)
else:
    print("Invalid input! Please enter a valid number.")
```

```python
import numpy as np
A = np.random.randint(0,20,(5,5))
print (A)
rank_A = np.linalg.matrix_rank(A)
print (rank_A)
if(rank_A == 5):
  print ("Matrix is invertible")
else:
  print("Matrix is not invertible")
det_A = np.linalg.det(A)
print (det_A)
if(det_A != 0):
  print("Matrix is invertible")
else:
  print("Matrix is not invertible")
```
